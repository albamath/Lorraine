# Rotation sur le cercle

```python
tau = 2 * RDF.pi()
angle = 2 * tau / 5
```

```python
a0 = 0.1*tau
increment = 0.4*tau
deb = point2d([cos(a0), sin(a0)], color='darkblue', size=10, zorder=15)
```

```python
depart = circle((0, 0), 1) + deb
```

```python
depart.show(axes=False, figsize=2)
```

```python
depart.save("point_sur_cercle.png", axes=False, figsize=3)
```

```python
def itere(k):
    a = angle_deb + k*increment
    p = cos(a), sin(a)
    C = circle((0, 0), 1)
    return C + point2d([p], color='red', size=50, zorder=12) + deb
```

```python
for k in range(6):
    itere(k).show(axes=False, figsize=2)
```

```python
for k in range(6):
    itere(k).save(f"itere_{k}.png", axes=False, figsize=3)
```

```python
def itere_rotation(k):
    a = angle_deb
    x = 2.5*k
    b = a + k*increment
    p = x + cos(a), sin(a)
    q = x + cos(b), sin(b)
    c = circle((x, 0), 1)
    dep = point2d([p], color='darkblue', size=20, zorder=15)
    arr = point2d([q], color='red', size=50, zorder=12)
    return c + dep + arr
```

```python
iteres = sum(itere_rotation(k) for k in range(6))
```

```python
iteres.show(axes=False)
```
